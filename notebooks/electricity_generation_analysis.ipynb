# Global Electricity Generation Analysis

## Project Overview

This project analyses global electricity generation patterns from 1900 to 2025 using data from Our World in Data (OWID).

The analysis explores how electricity generation has evolved over time, the contribution of different energy sources, and the differences in electricity generation and energy mix across selected countries.

## Key Questions

1. How has global electricity generation changed over time?
2. What proportion of global electricity comes from fossil fuels and low-carbon sources?
3. How quickly are solar and wind electricity generation growing?
4. How do electricity generation patterns differ across selected countries?
5. How has the electricity generation mix changed between 2000 and 2025?

## Data Source

Our World in Data – Energy Dataset

[OWID Energy Data](https://ourworldindata.org/energy)
